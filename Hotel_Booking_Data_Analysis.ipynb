```python
missing_values=df.isnull().sum()
missing_columns=missing_values[missing_values>0]
print(missing_columns)
```

```python
df['children'] = df['children'].fillna(df['children'].mode()[0])
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['agent'] = df['agent'].fillna(df['agent'].mean())
missing_values=df.isnull().sum()
missing_columns=missing_values[missing_values>0]
print(missing_columns)
```

```python
df = df.drop('company',axis=1)
df.isnull().sum()
```

```python
cat=df[['hotel','is_canceled','arrival_date_year','arrival_date_month',
        'meal','country','market_segment','distribution_channel',
        'is_repeated_guest','deposit_type','customer_type','reservation_status']]
cat
```

```python
num = df.select_dtypes(include=['int64','float64'])
num
```

```python
num.corr()
#correlation HeatMAP
plt.figure(figsize=(10,8))
sns.heatmap(num.corr(), annot=True,cmap='coolwarm')
plt.title('Correlation HeatMap')
plt.show()


```

```python
#count plot for categorical features
sns.countplot(x='hotel',data=df)
plt.title('Hotel Type Distribution')
plt.show()
```

```python
#Distribution of Numerical Features
sns.histplot(df['lead_time'], bins=30 ,kde=True)
plt.title('Lead Time Distribution')
plt.show()
```

```python
#Box plot to identify Outliers
num.plot(kind='box',subplots=True,layout=(4,5),figsize=(18,12))
plt.suptitle('Box plot for Numeric Columns')
plt.show()
```

```python
#pairplot for Numerical Variable
sns.pairplot(df[['lead_time','adr','stays_in_weekend_nights','is_canceled']], hue='is_canceled')
plt.show()
```

```python
#box plot to identify outliers
sns.boxplot(x='is_canceled',y='adr',data=df)
plt.title('ADR by Cancellation Status')
plt.show()
```
